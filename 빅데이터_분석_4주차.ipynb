#배열 결합 연습

```python
a= np.arange(1,4)
b= np.arange(4,7)
c= np.arange(7,10)
a,b,c
```

```python
p= np.arange(1,7).reshape(2,3)
q= np.arange(11,17).reshape(2,3)
r= np.arange(21,27).reshape(2,3)
p,q,r
```

```python
np.vstack((a,b,c)) # 10 --> 20
```

```python
np.vstack((p,q,r))  # 20 -->20
```

```python
np.hstack((a,b,c)) # 10 --> 10
```

```python
np.hstack((p,q,r)) # 20 --> 20
```

```python
np.column_stack((a,b,c)) # 10 --> 20
```

```python
np.column_stack((p,q,r)) # 20 --> 20
```

```python
np.concatenate((a,b,c))
```

```python
np.concatenate((a,b,c),axis=1)
```

```python
np.concatenate((a,b,c),axis=None)
```

```python
np.concatenate((a,b,c),axis = -1)
```

```python
np.concatenate((p,q,r))
```

```python
np.concatenate((p,q,r),axis=0)
```
